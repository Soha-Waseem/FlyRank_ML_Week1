## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target for this project is **`is_declining_label`**, a binary label that indicates whether a content item is declining in performance.

- **Target:** `is_declining_label`
- **Type:** Binary classification (0 = Not Declining, 1 = Declining)

## 3. Success metric

The primary success metric is ROC-AUC because this is a binary classification problem. ROC-AUC measures how well the model distinguishes between declining and non-declining content across different decision thresholds.

As a target, a ROC-AUC of 0.80 or higher would indicate good model performance, meaning the model can reliably distinguish between the two classes. A ROC-AUC of 0.50 indicates performance no better than random guessing, while 1.00 represents perfect classification.

## 5. Why ML beats a fixed rule here

A fixed rule, such as "if CTR is below 1%, then the content is declining," is too simplistic because content performance depends on many interacting factors rather than a single metric.

Each content item has multiple characteristics, including search volume, impressions, clicks, sessions, CTR, average search position, engagement rate, content age, and content type. These features can influence performance differently depending on the context.

A machine learning model can learn patterns from many variables simultaneously and identify relationships that are difficult to capture with manually defined rules. This allows content teams to make better-informed decisions about which content should be reviewed or refreshed.
